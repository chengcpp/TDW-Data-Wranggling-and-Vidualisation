# Introduction

Data wrangling is a process that involves handling and optimizing complex datasets to make them more accessible and analyzable. This process includes reorganizing, transforming, and mapping raw data into more structured and valuable information, thereby providing a foundation for various subsequent applications. For example, in sports analytics, football organizations can use data wrangling to organize match results, player statistics, and tournament outcomes. By cleaning and standardizing data from international football matches spanning decades, analysts can predict team performances, identify trends in match outcomes, and support decision-making for team strategies. Data wrangling also helps data scientists and analysts handle and analyze data more efficiently, providing more accurate support for decision-making.

As the demand for data processing continues to grow, data wrangling tools are playing an increasingly important role in data management processes. These tools not only transform unprocessed raw data into standardized formats but also effectively remove noise or missing information from the data, ensuring the accuracy of the analysis process. By identifying and addressing errors, anomalies, and inconsistencies in the data, data wrangling significantly improves the reliability and quality of the data. Additionally, it greatly simplifies the management of large datasets, making data flow and sharing more seamless and providing strong support for various analyses and business processes.

2.1.5 Disadvantages of EDA In Healthcare

Despite its advantages, EDA also presents some challenges.

Data quality is a critical concern, as EDA is highly dependent on the accuracy and completeness of the input data. Poor-quality data can lead to incorrect conclusions, which is especially dangerous in a healthcare setting (Shirzadfar et al., 2018).

The preprocessing phase of EDA, which involves cleaning and organizing large datasets, can be time-consuming and may require significant computational resources (Gubbi et al., 2013).

While EDA is valuable for exploring data and generating hypotheses, it does not inherently provide predictive power. For accurate predictions, further statistical or machine learning models are required (Swamynathan, 2017).

Additionally, EDA requires a certain level of expertise, as proper interpretation of the results requires knowledge of both healthcare and data analysis, making it less accessible to those without technical training (Patel et al., 2012).


2.2 EDA in Manufacturing

Exploratory Data Analysis (EDA) is an essential first step for interpreting data before applying more complex techniques like predictive modeling. In smart manufacturing, EDA is used to uncover patterns, trends, and abnormalities that influence production efficiency, machine reliability, and overall performance in manufacturing systems. With Industry 4.0 technologies generating massive amounts of data through sensors and IoT devices, EDA becomes vital for turning raw data into valuable insights.

EDA plays a crucial role in analyzing sensor and ERP (Enterprise Resource Planning) data from Dolle A/S, a wooden loft ladder manufacturer. Through EDA, the authors seek to identify patterns responsible for machine downtime, production slowdowns, and other inefficiencies in the manufacturing workflow. This review looks at the role of EDA in smart manufacturing, drawing from the case study and related academic research.

2.2.1 Problems Faced in Manufacturing

1. Smart Manufacturing

One of the major issues addressed through Exploratory Data Analysis (EDA) in manufacturing is machine downtime. By leveraging sensor data, EDA helped identify patterns that contributed to frequent machine stoppages, such as errors in the screw machines or defects in the strings. This insight was key to minimizing unplanned shutdowns and boosting overall equipment efficiency. EDA was also instrumental in identifying production delays. By analyzing variables like PaceIn (the rate at which materials enter the machine) and PaceOut (the speed at which finished products are produced), EDA revealed factors causing slowdowns. Equipment malfunctions, particularly screw machine errors, were identified as major contributors to these delays. Another challenge in manufacturing data was its skewness and uneven distribution. For instance, MachineOn showed extreme skewness, being active 98% of the time. EDA helped detect such irregularities, allowing for the correction of skewed variables, which otherwise could have led to misinterpretation of the data. Additionally, EDA revealed important correlations between machine failures, faulty components, and downtime. Through the use of multivariate analysis and correlation matrices, relationships between factors like faulty strings and screw errors were discovered, providing critical insights into the root causes of equipment malfunctions and informing preventive strategies. Finally, EDA tackled issues related to data granularity. The sensor data was recorded at one-second intervals, but multiple changes sometimes occurred within that period, resulting in missing or incomplete information. EDA helped aggregate this data, allowing for the identification of larger patterns and ensuring that these gaps did not lead to erroneous conclusions.


2. Pharmaceutical Manufacturing

The fluid bed dryer faced several problems that hindered its efficiency. The preheating phase consumed more energy than necessary, leading to high energy consumption. 2.2.5 Disadvantages of EDA In Manufacturing

One major challenge of conducting EDA in manufacturing settings is the handling of large, intricate datasets. Manufacturing data typically includes a combination of sensor readings, alarms, and machine logs, which are often skewed or contain missing values. As Iftikhar et al. (2019) noted, sensor data, particularly in binary format, tends to exhibit skewness. Variables like FaultyString and ScrewError often have extreme distributions, making traditional EDA methods less effective without proper data cleaning and preprocessing.

Another challenge is data granularity. In the Dolle case study, sensor data was logged at one-second intervals, but in some cases, sensors may change state multiple times within that interval, leaving potential gaps. 3.5 Detecting and Visualizing Outliers Using Z-Score

To detect outliers in football match scores, focusing on both home and away teams. The box plots show the presence of extreme outliers, particularly in high-scoring matches, where home teams tend to have more significant score anomalies compared to away teams.

```python
# Function to detect outliers using Z-score
def detect_outliers_zscore(df, column):
    z_scores = stats.zscore(df[column].dropna())  # Calculate Z-scores, excluding NaN values
    abs_z_scores = np.abs(z_scores)  # Take absolute values of Z-scores
    outliers = df[abs_z_scores > 3]  # Using Z-score threshold of 3 to detect outliers
    return outliers


# Convert columns to numeric where applicable and handle non-numeric values by coercing them to NaN
goalscorers['minute'] = pd.to_numeric(goalscorers['minute'], errors='coerce')
results['home_score'] = pd.to_numeric(results['home_score'], errors='coerce')
results['away_score'] = pd.to_numeric(results['away_score'], errors='coerce')

# Reapply the Z-score method for outlier detection
minute_outliers_z = detect_outliers_zscore(goalscorers.dropna(subset=['minute']), 'minute')
home_score_outliers_z = detect_outliers_zscore(results, 'home_score')
away_score_outliers_z = detect_outliers_zscore(results, 'away_score')

# Display outliers detected
display(minute_outliers_z, home_score_outliers_z, away_score_outliers_z)

# Box plot for Home Score
plt.subplot(1, 2, 1)
plt.boxplot(results['home_score'].dropna(), vert=False)
plt.title('Home Score Box Plot')
plt.xlabel('Home Score')

# Box plot for Away Score
plt.subplot(1, 2, 2)
plt.boxplot(results['away_score'].dropna(), vert=False)
plt.title('Away Score Box Plot')
plt.xlabel('Away Score')

plt.tight_layout()
plt.show()
```

4.10 Identifying the First Goal Scorer in Each Match

```python
#Identify the first goal scorer in each match, filtering out "Unknown" scorers
detailed_df['minute'] = pd.to_numeric(detailed_df['minute'], errors='coerce')

# Filter rows where a goal was scored and the scorer is not "Unknown"
goals_scored = detailed_df.dropna(subset=['scorer'])
goals_scored = goals_scored[goals_scored['scorer'] != 'Unknown']
goals_scored = goals_scored[goals_scored['minute'] != 'Unknown']

# Sort the goals by match and time (minute)
goals_scored_sorted = goals_scored.sort_values(by=['date', 'home_team', 'away_team', 'minute'])

# Group by each match and identify the first goal scorer
first_goal_scored = goals_scored_sorted.groupby(['date', 'home_team', 'away_team']).first().reset_index()

# Add a column to indicate which team scored first (home or away team)
first_goal_scored['first_goal_team'] = first_goal_scored['team']
```

4.11 Categorizing the Timing of the First Goal

To categorize the first goal in each match based on when it was scored, dividing it into early, mid-game, or late goals. The categorization is merged with the match-level data to ensure the timing is properly categorized

```python
# Categorize the timing of the first goal, handle NaN values in 'minute'
def categorize_goal_timing(minute):
    if pd.isna(minute):
        return 'Unknown'
    elif minute <= 30:
        return 'Early Goal (0-30 min)'
    elif minute <= 60:
        return 'Mid-Game Goal (31-60 min)'
    else:
        return 'Late Goal (61+ min)'

# Ensure 'minute' column is present and apply the categorisation to the first goal scored dataframe
if 'minute' in first_goal_scored.columns:
    first_goal_scored['goal_timing_category'] = first_goal_scored['minute'].apply(categorize_goal_timing)
    first_goal_scored = first_goal_scored[first_goal_scored['goal_timing_category'] != 'Unknown']

# Merge the first goal data with match-level data, including 'goal_timing_category'
if 'goal_timing_category' in first_goal_scored.columns:
    merged_first_goal = match_level_df.merge(
        first_goal_scored[['date', 'home_team', 'away_team', 'first_goal_team', 'minute', 'goal_timing_category']],
        on=['date', 'home_team', 'away_team'],
        how='inner'
    )
else:
    print("Error: 'goal_timing_category' was not created properly.")
```

4.12 Analyzing Win Rates When Scoring the First Goal

To calculate the win rate of teams that scored the first goal and examines how goal timing affects the win rate. It groups matches by goal timing categories (early, mid-game, late) to determine how timing influences a team's chances of winning when they score first.

```python
# Check for missing 'first_goal_team' and drop them
merged_first_goal = merged_first_goal.dropna(subset=['first_goal_team'])

# Check if the team that scored first went on to win
first_goal_wins = (merged_first_goal['first_goal_team'] == merged_first_goal['home_team']) & (merged_first_goal['Result'] == 'Home Win')
first_goal_wins = first_goal_wins | ((merged_first_goal['first_goal_team'] == merged_first_goal['away_team']) & (merged_first_goal['Result']
                                                                                                                 == 'Away Win'))

# Calculate the total matches and win rate when scoring first
total_first_goal_matches = merged_first_goal.shape[0]
first_goal_win_rate = first_goal_wins.sum() / total_first_goal_matches * 100 if total_first_goal_matches > 0 else 0

# Calculate win rates for each goal timing category, filter out 'Unknown'
merged_first_goal = merged_first_goal[merged_first_goal['goal_timing_category'] != 'Unknown']
merged_first_goal['first_goal_team_won'] = first_goal_wins

# Group by goal timing category and calculate the win rate for each category
win_rate_by_goal_timing = merged_first_goal.groupby('goal_timing_category')['first_goal_team_won'].mean() * 100

# Print the results
print(f"Teams' win rate when scoring first: {first_goal_win_rate:.2f}%")
print("Win Rate by Goal Timing:")
print(win_rate_by_goal_timing)
```

4.13 Calculating Equalization and Comeback Rates by Goal Timing

To calculates the equalization and comeback rates based on when the first goal was scored. It categorizes matches by goal timing and computes how often teams equalize or come from behind to win, providing insights into match dynamics across different goal timings

```python
# Now proceed with calculating equalization and comeback rates
# Add columns for 'equalized' and 'came_from_behind_and_won'
merged_first_goal['equalized'] = merged_first_goal.apply(
    lambda row: (row['Result'] == 'Draw') or
                ((row['first_goal_team'] == row['home_team']) and (row['Result'] == 'Away Win')) or
                ((row['first_goal_team'] == row['away_team']) and (row['Result'] == 'Home Win')),
    axis=1
)

merged_first_goal['came_from_behind_and_won'] = merged_first_goal.apply(
    lambda row: ((row['first_goal_team'] == row['home_team']) and (row['Result'] == 'Away Win')) or
                ((row['first_goal_team'] == row['away_team']) and (row['Result'] == 'Home Win')),
    axis=1
)

# Group by goal timing category and calculate equalization and comeback rates
grouped_by_goal_timing = merged_first_goal.groupby('goal_timing_category').agg(
    total_trailing_matches=('equalized', 'count'),
    total_equalized=('equalized', 'sum'),
    total_came_from_behind_and_won=('came_from_behind_and_won', 'sum')
)

# Calculate rates by goal timing category
grouped_by_goal_timing['equalization_rate'] = (grouped_by_goal_timing['total_equalized'] / grouped_by_goal_timing['total_trailing_matches']) * 100
grouped_by_goal_timing['comeback_rate'] = (grouped_by_goal_timing['total_came_from_behind_and_won'] / grouped_by_goal_timing
                                           ['total_trailing_matches'])* 100

# Print the final results
print(grouped_by_goal_timing[['equalization_rate', 'comeback_rate']])
```

To examine the outcomes of football matches in terms of draws, low-scoring matches, and wins by a small margin (1-0 or 2-1) in major versus other tournaments. By comparing the percentages of these outcomes, the code provides insights into the differences in match competitiveness and scoring patterns across tournament types.

```python
# Wins by a Small Margin (1-0 or 2-1) Calculation
match_level_df.loc[:, 'small_margin_win'] = ((match_level_df['home_score'] == 1) & (match_level_df['away_score'] == 0)) | \
                                            ((match_level_df['home_score'] == 2) & (match_level_df['away_score'] == 1)) | \
                                            ((match_level_df['home_score'] == 0) & (match_level_df['away_score'] == 1)) | \
                                            ((match_level_df['home_score'] == 1) & (match_level_df['away_score'] == 2))

# Calculate the percentage of small-margin wins for major vs. other tournaments
small_margin_comparison = match_level_df.groupby('is_major_tournament').agg(
    total_small_margin=('small_margin_win', 'sum'),
    total_matches=('is_major_tournament', 'count')
)
small_margin_comparison['small_margin_percentage'] = (small_margin_comparison['total_small_margin'] / small_margin_comparison['total_matches']) * 100

# Print the results
print("Draw Percentage Comparison In Regular Time (Major vs. Other Tournaments):")
print(draw_percentage_comparison[['draw_percentage']])

print("\nLow-Scoring Matches (<2 Goals) Percentage (Major vs. Other Tournaments):")
print(low_scoring_comparison[['low_scoring_percentage']])

print("\nSmall-Margin Wins (1-0 or 2-1) Percentage (Major vs. Other Tournaments):")
print(small_margin_comparison[['small_margin_percentage']])
```

4.18 Percentage of Matches Ending in Shootouts in Major vs. Other Tournaments

To compare the percentage of football matches that ended in shootouts between major tournaments and other tournaments. The code calculates and reveals that shootouts are more frequent in major tournaments than in other tournaments, providing insights into match intensity and decision-making processes in different types of tournaments

```python
# Perform the aggregation using the temporary column
shootout_percentage_comparison = match_level_df.groupby('is_major_tournament').agg(
    total_shootouts=('had_shootout', 'sum'),  # Use the temporary column
    total_matches=('is_major_tournament', 'count')
)

# Calculate the percentage of matches that ended in a shootout
shootout_percentage_comparison['shootout_percentage'] = (shootout_percentage_comparison['total_shootouts'] / shootout_percentage_comparison
                                                         ['total_matches']) * 100

# Print the result
print("Percentage of Matches Ending in Shootouts (Major vs. Other Tournaments):")
print(shootout_percentage_comparison[['shootout_percentage']])
```

4.20 Filtering Top Scorers by Average Goals per Match

To filters out "Unknown" scorers and calculates the average goals per match for each valid scorer. It identifies the top 10% of scorers based on average goals per match and displays them. The output ranks players by their total goals and matches played, providing insights into their scoring efficiency.

```python
# Filter out "Unknown" scorers and calculate total and average goals per match
valid_scorers = detailed_df[detailed_df['scorer'] != 'Unknown']

# Group by scorer and the combination of date, home_team, and away_team to calculate the actual number of matches
scorer_stats = valid_scorers.groupby('scorer').agg(
    total_goals=('scorer', 'size'),
    num_matches=('date', 'nunique')
)

# Calculate the average goals per match for each scorer
scorer_stats['avg_goals_per_match'] = scorer_stats['total_goals'] / scorer_stats['num_matches']

# Determine the threshold for the top 10% of scorers by average goals per match
top_10_percent_threshold = scorer_stats['avg_goals_per_match'].quantile(0.90)

# Filter the dataset for only the top 10% scorers based on average goals per match
top_scorers_by_avg_goals = scorer_stats[scorer_stats['avg_goals_per_match'] >= top_10_percent_threshold]

# Filter the dataset for matches where the top scorers played
top_scorer_matches = valid_scorers[valid_scorers['scorer'].isin(top_scorers_by_avg_goals.index)]

# Display top scorers
print("Top 10% Scorers (Excluding 'Unknown') ranked by average goals per match:")
print(top_scorers_by_avg_goals)
```

4.21 Analysis of Win Rates for Teams with Top Scorers Across Major and Regular Tournaments

To investigate the impact of top scorers on team performance across major and regular tournaments. By classifying teams based on the presence of top scorers and calculating win rates, the analysis reveals how teams with high-performing scorers fare in both major and regular competitions. The goal is to identify if having a top scorer correlates with a higher chance of winning

```python
# Classify teams based on whether they have top scorers
top_scorer_team_wins = top_scorer_matches.groupby(['team']).size()

# Merge with the main match dataset to analyze team wins
team_with_top_scorers = match_level_df.merge(top_scorer_team_wins.rename('top_scorer_goals'), left_on='home_team', right_index=True, how='left').fillna(0)

# Calculate win rates of teams with top scorers in major vs regular tournaments
# Add a column for tournament type
major_tournaments = ['FIFA World Cup', 'UEFA Euro', 'Copa América', 'African Cup of Nations', 'AFC Asian Cup', 'Gold Cup']  # Define other major tournaments
team_with_top_scorers['tournament_type'] = team_with_top_scorers['tournament'].apply(lambda x: 'Major' if x in major_tournaments else 'Regular')

# Calculate win rates by tournament type (filter matches where top scorers were present)
win_rate_by_tournament_type = team_with_top_scorers[team_with_top_scorers['top_scorer_goals'] > 0].groupby('tournament_type')['Result'].apply(lambda x: (x == 'Home Win').mean() * 100)

# Display the win rates for teams with top scorers
print("Win Rates for Teams with Top Scorers by Tournament Type:")
print(win_rate_by_tournament_type)
```

4.22 Contribution of Top Scorers to Team Wins by Tournament Type

To explore how top scorers contribute to team victories across various tournaments. By calculating win rates and top scorer contributions, the results show how crucial top scorers are in major tournaments like the FIFA World Cup and UEFA Euro compared to regular tournaments

```python
# Analyze the contribution of top scorers to team wins
# Calculate the percentage of total goals scored by top scorers for each match, handling division by zero
team_with_top_scorers['total_goals'] = team_with_top_scorers['home_score'] + team_with_top_scorers['away_score']
team_with_top_scorers['top_scorer_goal_contrib'] = team_with_top_scorers.apply(lambda row: row['top_scorer_goals'] / row['total_goals'] if row['total_goals'] > 0 else 0, axis=1)

# Calculate the contribution in major tournaments (filter matches where top scorers were present)
contribution_by_tournament_type = team_with_top_scorers[team_with_top_scorers['top_scorer_goals'] > 0].groupby('tournament_type')['top_scorer_goal_contrib'].mean()

# Display the contribution of top scorers to team wins
print("Contribution of Top Scorers to Wins by Tournament Type:")
print(contribution_by_tournament_type)

# Convert win_rate_by_tournament_type to a DataFrame
win_rate_df = win_rate_by_tournament_type.reset_index()
win_rate_df.columns = ['Tournament Type', 'Win Rate (%)']

# Convert contribution_by_tournament_type to a DataFrame
contribution_df = contribution_by_tournament_type.reset_index()
contribution_df.columns = ['Tournament Type', 'Contribution (%)']
```
